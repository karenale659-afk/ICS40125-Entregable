### **Número de títulos agregados por año**

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Convert 'date_added' to datetime objects, handling mixed formats
df['date_added'] = pd.to_datetime(df['date_added'], format='mixed', dayfirst=False)

# Extract the year from 'date_added'
df['year_added'] = df['date_added'].dt.year

# Count the number of titles added each year
titles_added_per_year = df['year_added'].value_counts().sort_index()

# Plotting the time series
plt.figure(figsize=(12, 6))
sns.lineplot(x=titles_added_per_year.index, y=titles_added_per_year.values, marker='o')
plt.title('Número de Títulos Agregados a Netflix por Año')
plt.xlabel('Año de Incorporación')
plt.ylabel('Cantidad de Títulos Agregados')
plt.grid(True, linestyle='--', alpha=0.7)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

**¿Cuándo creció más el catálogo de Netflix? ¿Qué relación podría tener con la expansión internacional de la plataforma?**

Analizando el gráfico, el catálogo de Netflix experimentó su mayor crecimiento en los años recientes, con un incremento significativo de títulos añadidos anualmente, especialmente en la década de 2010 y principios de 2020. Este crecimiento coincide directamente con la expansión internacional agresiva de Netflix, que comenzó a escalar globalmente alrededor de 2010-2015, buscando no solo nuevos suscriptores sino también produciendo y adquiriendo contenido local para diversas regiones.

### **Heatmap de lanzamientos por año y mes**

```python
import calendar

# Extract month from 'date_added'
df['month_added'] = df['date_added'].dt.month

# Create a pivot table for the heatmap
heatmap_data = df.groupby(['year_added', 'month_added']).size().unstack(fill_value=0)

# Reorder columns to represent months correctly
heatmap_data = heatmap_data[list(range(1, 13))]

# Map month numbers to month names for better readability
month_names = [calendar.month_abbr[i] for i in range(1, 13)]
heatmap_data.columns = month_names

plt.figure(figsize=(14, 8))
sns.heatmap(heatmap_data, cmap='viridis', annot=True, fmt='d', linewidths=.5)
plt.title('Número de Títulos Agregados a Netflix por Año y Mes')
plt.xlabel('Mes de Incorporación')
plt.ylabel('Año de Incorporación')
plt.yticks(rotation=0)
plt.show()
```

**¿Hay meses con más estrenos que otros? ¿Qué conclusiones puedes sacar sobre la estacionalidad de lanzamientos?**

El heatmap revela una clara estacionalidad en la adición de títulos a Netflix. Generalmente, los meses de **diciembre, enero, julio y octubre** muestran un mayor número de estrenos o incorporaciones de contenido. Esto podría deberse a varias razones:

*   **Temporadas de vacaciones (diciembre/enero):** Las plataformas de streaming suelen lanzar mucho contenido nuevo durante las vacaciones de invierno y fin de año para capitalizar el tiempo libre de los suscriptores y atraer nuevas audiencias.
*   **Verano (julio):** De manera similar, los meses de verano pueden ser un buen momento para lanzar contenido, ya que muchas personas tienen más tiempo libre.
*   **Cambios de temporada (octubre):** Octubre se asocia a menudo con lanzamientos de contenido más oscuro o de terror para Halloween, y también marca el inicio de la temporada de premios, lo que puede influir en la programación.

En contraste, meses como **febrero, marzo y septiembre** a menudo muestran un volumen de lanzamientos ligeramente menor, aunque no de forma drástica. Esta estacionalidad sugiere una estrategia de contenido bien planificada por parte de Netflix para maximizar la audiencia en períodos clave del año.

### **Duración de películas por género**

```python
# Filtrar solo películas
df_movies = df[df['type'] == 'Movie'].copy()

# Extraer la duración en minutos
df_movies['duration_minutes'] = df_movies['duration'].str.extract('(\d+) min').astype(float)

# Extraer el género principal (el primero listado)
df_movies['main_genre'] = df_movies['listed_in'].apply(lambda x: x.split(',')[0].strip())

# Limpiar datos nulos de duración
df_movies.dropna(subset=['duration_minutes'], inplace=True)

# Identificar los top N géneros para una mejor visualización
top_genres = df_movies['main_genre'].value_counts().nlargest(10).index
df_movies_top_genres = df_movies[df_movies['main_genre'].isin(top_genres)]

# Ordenar los géneros por la mediana de duración para la visualización
sorted_genres = df_movies_top_genres.groupby('main_genre')['duration_minutes'].median().sort_values(ascending=False).index

plt.figure(figsize=(14, 8))
sns.boxplot(data=df_movies_top_genres, x='main_genre', y='duration_minutes', order=sorted_genres, palette='viridis')
plt.title('Distribución de Duración de Películas por Género Principal (Top 10)')
plt.xlabel('Género Principal')
plt.ylabel('Duración (minutos)')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

**¿Qué géneros suelen durar más? ¿Qué podrías decir sobre la variabilidad dentro de un mismo género?**

El boxplot muestra que géneros como las películas de acción y aventura, los dramas y las comedias tienden a tener duraciones promedio más largas. Por otro lado, géneros como las películas infantiles y familiares o los documentales pueden mostrar duraciones más cortas o una mayor variabilidad.

En cuanto a la variabilidad:
*   Dramas y Películas de acción y aventura a menudo presentan una alta variabilidad, con películas que van desde duraciones estándar hasta algunas mucho más largas (outliers superiores), lo que es común en producciones épicas o con tramas complejas.
*   Otros géneros, como las películas infantiles, pueden tener una distribución de duración más compacta, lo que indica un rango más consistente en sus tiempos de ejecución. La presencia de outliers en algunos géneros sugiere que, aunque hay una duración típica, existen excepciones significativas (películas muy cortas o muy largas) que se desvían de la norma.

### **Combinaciones más frecuentes de Género y Clasificación por Edad**

```python
# Extraer el género principal para todo el DataFrame
df['main_genre'] = df['listed_in'].apply(lambda x: x.split(',')[0].strip())

# Eliminar filas con valores nulos en 'rating' o 'main_genre' para el análisis
df_filtered = df.dropna(subset=['rating', 'main_genre']).copy()

# Agrupar por género principal y rating, y contar las ocurrencias
genre_rating_counts = df_filtered.groupby(['main_genre', 'rating']).size().reset_index(name='count')

# Identificar los top N géneros y ratings para una visualización más clara
top_n_genres = genre_rating_counts['main_genre'].value_counts().nlargest(15).index
top_n_ratings = genre_rating_counts['rating'].value_counts().nlargest(10).index

# Filtrar el DataFrame para incluir solo los top géneros y ratings
genre_rating_filtered = genre_rating_counts[
    (genre_rating_counts['main_genre'].isin(top_n_genres)) &
    (genre_rating_counts['rating'].isin(top_n_ratings))
]

# Crear una tabla pivote para el heatmap
heatmap_pivot = genre_rating_filtered.pivot_table(index='main_genre', columns='rating', values='count', fill_value=0)

# Ordenar los géneros por la suma total de títulos para una mejor lectura
heatmap_pivot = heatmap_pivot.loc[heatmap_pivot.sum(axis=1).sort_values(ascending=False).index]

# Reordenar las columnas de rating en un orden lógico (ej. de infantil a adulto, si es posible)
# Los ratings disponibles en el kernel state son: 'TV-MA', 'TV-14', 'TV-PG', 'R', 'PG-13', 'TV-Y7', 'TV-Y', 'PG', 'TV-G', 'NR', 'G', 'TV-Y7-FV', 'NC-17', 'UR'
# Intentamos un orden aproximado:
rating_order = ['TV-Y', 'TV-Y7', 'TV-Y7-FV', 'TV-G', 'PG', 'PG-13', 'TV-PG', 'TV-14', 'R', 'TV-MA', 'NC-17', 'NR', 'UR']
heatmap_pivot = heatmap_pivot.reindex(columns=[r for r in rating_order if r in heatmap_pivot.columns], fill_value=0)

plt.figure(figsize=(16, 10))
sns.heatmap(heatmap_pivot, cmap='YlGnBu', annot=True, fmt='.0f', linewidths=.5, linecolor='black')
plt.title('Heatmap de Frecuencia de Género vs. Clasificación por Edad')
plt.xlabel('Clasificación por Edad')
plt.ylabel('Género Principal')
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```

**¿Qué géneros están más dirigidos a un público adulto? ¿Y cuáles a público familiar o infantil?**

El heatmap ofrece una visión clara de la relación entre géneros y clasificaciones por edad:

*   **Géneros dirigidos a público adulto (TV-MA, R, TV-14):** Los Dramas, las Películas de acción y aventura, las Comedias y los Thrillers muestran una fuerte concentración en clasificaciones como TV-MA, R y TV-14. Esto sugiere que Netflix produce y adquiere principalmente contenido adulto en estos géneros.

*   **Géneros dirigidos a público familiar o infantil (TV-Y, TV-Y7, TV-G, PG):** Las Películas infantiles y familiares (Children & Family Movies), las Películas para niños (Kids' TV) y las Series documentales (Docuseries) dominan las clasificaciones TV-Y, TV-Y7, TV-G y PG. Esto es esperado, ya que son géneros intrínsecamente diseñados para audiencias más jóvenes.

*   **Géneros mixtos o con menor predominancia:** Algunos géneros, como los Documentales, pueden tener una distribución más amplia de ratings, aunque con una inclinación hacia lo adulto. Otros, como los Anime Features o Sci-Fi & Fantasy, pueden mostrar patrones más variados dependiendo del contenido específico.

En general, Netflix parece tener una estrategia de contenido bien definida, donde géneros específicos se asocian consistentemente con rangos de edad particulares.
